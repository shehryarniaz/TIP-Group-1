```python
#normalization 
from sklearn.preprocessing import MinMaxScaler
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns
scaler = MinMaxScaler()
df[numerical_features] = scaler.fit_transform(df[numerical_features])
print(df.head())
```

```python
from sklearn.preprocessing import LabelEncoder
categorical_features = df.select_dtypes(include=['object']).columns
#Label Encoding
label_encoders = {}
for col in categorical_features:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

# Print the dataframe with label encoded columns
print(df.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Generate sample data
np.random.seed(42)
data = pd.DataFrame({
    'int_column': np.random.randint(1, 100, 100),  # Integer values
    'object_column': np.random.uniform(1.0, 100.0, 100)  # Float values
})

# Line Plot
plt.figure(figsize=(10, 6))
plt.plot(data.index, data['int_column'], label='Integer Column', marker='o')
plt.plot(data.index, data['object_column'], label='object Column', marker='x')
plt.xlabel('Index')
plt.ylabel('Values')
plt.title('Line Plot of Integer and object Columns')
plt.legend()
plt.show()

# Histogram
plt.figure(figsize=(10, 6))

# Histogram for integer values
plt.hist(data['int_column'], bins=20, alpha=0.5, label='Integer Column', edgecolor='black')

# Histogram for object values
plt.hist(data['object_column'], bins=20, alpha=0.5, label='Float Column', edgecolor='black')

plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram of Integer and Float Columns')
plt.legend()
plt.show()

# Box Plot
plt.figure(figsize=(10, 6))
sns.boxplot(data=data[['int_column', 'object_column']])
plt.ylabel('Values')
plt.title('Box Plot of Integer and Float Columns')
plt.show()

# Violin Plot
plt.figure(figsize=(10, 6))
sns.violinplot(data=data[['int_column', 'object_column']])
plt.ylabel('Values')
plt.title('Violin Plot of Integer and object Columns')
plt.show()

# Scatter Plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='int_column', y='object_column', data=data)
plt.xlabel('Integer Column')
plt.ylabel('Float Column')
plt.title('Scatter Plot of Integer and Float Columns')
plt.show()

# Plotly Scatter Plot
fig = px.scatter(data, x='int_column', y='float_column', title='Scatter Plot of Integer and Float Columns')
fig.show()

# Plotly Histogram for Integer Column
fig = px.histogram(data, x='int_column', nbins=20, title='Histogram of Integer Column')
fig.show()

# Plotly Histogram for Float Column
fig = px.histogram(data, x='float_column', nbins=20, title='Histogram of Float Column')
fig.show()
```
